```python
from syntethic_causal_data import SyntheticCausalSystem
```

```python
gen_linear_no_conf=SyntheticCausalSystem(n_features=10,
                                random_state=42,
                                min_num_connected_edges=2,
                                edge_probability=0.3)
data, adj, conf = gen_linear_no_conf.generate_linear_system(n_samples=1000,
                                                    with_confounders=False,
                                                    noise_std=0.5,
                                                    y_parents_ratio=0.4)
plt_obj = gen_linear_no_conf.visualize_causal_graph(
    adj,
    conf,
    title=' Linear system without confounders with intermidite edge probability',
    y_parent_indices=gen_linear_no_conf.y_parent_indices

)
```
